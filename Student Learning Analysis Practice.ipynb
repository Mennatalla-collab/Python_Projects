## FILLNA ##

```python
learning_activity["Lessons_Completed"] = learning_activity["Lessons_Completed"].fillna(learning_activity["Lessons_Completed"].median())

learning_activity["Quiz_Score"] = learning_activity["Quiz_Score"].fillna(
    learning_activity["Quiz_Score"].median()
)

learning_activity["Assignment_Score"] = learning_activity["Assignment_Score"].fillna(
    learning_activity["Assignment_Score"].median()
)

learning_activity["Study_Time_Minutes"] = learning_activity["Study_Time_Minutes"].fillna(
    learning_activity["Study_Time_Minutes"].median()
)

learning_activity["Completion_Percentage"] = learning_activity["Completion_Percentage"].fillna(
    learning_activity["Completion_Percentage"].median()
)

learning_activity["Platform_Rating"] = learning_activity["Platform_Rating"].fillna(
    learning_activity["Platform_Rating"].median()
)
```

```python
learning_activity["Activity_Date"] = pd.to_datetime(learning_activity["Activity_Date"], errors='coerce')
```

## MERGE ##

```python
merged_data = pd.merge(student, learning_activity, on="Student_ID")

result = merged_data.groupby(
    ["Student_ID", "Student_Name", "City", "Learning_Plan"]
)["Study_Time_Minutes"].sum().reset_index()
result["Study_Time_Minutes"] = result["Study_Time_Minutes"].astype("Int64")
```

```python
result 
```

## Total Study Time by Course for Tanta Students ##

```python
result = pd.merge(
    student[["Student_ID", "City"]],
    learning_activity[["Student_ID", "Course", "Study_Time_Minutes"]],
    on="Student_ID",
    how="left"
)

result[result["City"] == "Tanta"].groupby("Course")["Study_Time_Minutes"].sum().reset_index()
```

## Average Quiz Score by Course for Bachelor Students ##

```python
result = pd.merge(
    student[["Student_ID", "Education_Level"]],
    learning_activity[["Student_ID", "Course", "Quiz_Score"]],
    on="Student_ID",
    how="left"
)

result[result["Education_Level"] == "Bachelor"].groupby("Course")["Quiz_Score"].mean().reset_index()
```

## Total Study Time by Learning Plan ##

```python
result = pd.merge(
    student[["Student_ID", "Learning_Plan"]],
    learning_activity[["Student_ID", "Study_Time_Minutes"]],
    on="Student_ID",
    how="left"
)

result.groupby("Learning_Plan")["Study_Time_Minutes"].sum().reset_index()
```

## Average Completion Percentage by City ##

```python
result = pd.merge(
    student[["Student_ID", "City"]],
    learning_activity[["Student_ID", "Completion_Percentage"]],
    on="Student_ID",
    how="left"
)

result.groupby("City")["Completion_Percentage"].mean().reset_index()
```

## Highest Quiz Score by Learning Plan ##

```python
result = pd.merge(
    student[["Student_ID", "Learning_Plan"]],
    learning_activity[["Student_ID", "Quiz_Score"]],
    on="Student_ID",
    how="left"
)

result.groupby("Learning_Plan")["Quiz_Score"].max().reset_index()
```

## Top 5 Students By Total Study Time ##

```python
top_students = (
    pd.merge(
        student[["Student_ID", "Student_Name"]],
        learning_activity[["Student_ID", "Study_Time_Minutes"]],
        on="Student_ID",
        how="inner"
    )
    .groupby(["Student_ID", "Student_Name"])["Study_Time_Minutes"]
    .sum().reset_index()
    .sort_values(by="Study_Time_Minutes",ascending=False)
    .head(5)
)

print(top_students)
```
